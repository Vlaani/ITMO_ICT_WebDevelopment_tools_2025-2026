```python
import json

with open('example2.json') as f:
    d = json.load(f)
```

```python
product = {}

product['name'] = d['data']['result']['GLOBAL_DATA']['globalData']['subject']
product['attributes'] = []
for att in d['data']['result']['PRODUCT_PROP_PC']['showedProps']:
    product['attributes'].append(att)

product
```

```python
len(d['data']['result']['PRICE']['skuIdStrPriceInfoMap'])
```

```python
variants = {}

for key, val in d['data']['result']['PRICE']['skuIdStrPriceInfoMap'].items():
    variants[key] = {"price":float(val['salePriceString'][:-2].replace(" ", "").replace(',', '.'))}

variants
```

```python
props = {}

for prop in d['data']['result']['SKU']['skuProperties']:
    props[prop['skuPropertyId']] = {'attribute': prop['skuPropertyName'], "vals":{}}
    for val in prop['skuPropertyValues']:
        props[prop['skuPropertyId']]['vals'][val['propertyValueIdLong']] = val['propertyValueName'].lower()
    
props
```

```python
for path in d['data']['result']['SKU']['skuPaths']:
    print(path['path'].split(";"))
    variants[path['skuIdStr']]['stock'] = path['skuStock']
    variants[path['skuIdStr']]['stock'] = path['skuStock']
    
variants
```
